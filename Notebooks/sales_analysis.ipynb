Visualization

```python
# Bar Chart For: top products

import matplotlib.pyplot as plt

top_product=df.groupby('Product Name')["Sales"].sum().sort_values(ascending = False).head(10)

colors = ['red', 'blue', 'green', 'orange', 'purple',
          'pink', 'brown', 'cyan', 'magenta', 'yellow']

plt.figure(figsize=(60,20))

plt.bar(top_product.index,top_product.values,color=colors)

plt.title("Top 10 Selling Products",size=50,fontweight='bold')
plt.xlabel("Product Name",size=50,fontweight='bold')
plt.ylabel("Total Sales",size=50,fontweight='bold')
plt.style.use('default')
#plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Pie Chart For: category distribution

category=df.groupby('Category')["Sales"].sum()

colors=['skyblue', 'orange', 'lightgreen']

plt.figure(figsize=(8,8))

plt.pie(category,
        labels=category.index,
        autopct='%1.2f%%',
        colors=colors,
        shadow=True,
        startangle=90,
        explode=(0.05,0.05,0.20)
       )
plt.style.use('ggplot')
plt.style.use('default')
#plt.style.use('dark_background')
plt.title("Category Distribution",size=20,fontweight='bold')

plt.tight_layout()

plt.show()
```

```python
# Donut Chart

import matplotlib.pyplot as plt

# Category-wise sales
category = df.groupby('Category')["Sales"].sum()

# Colors
#colors = ['skyblue', 'orange', 'lightgreen']
colors = ['#F28E2B','#E15759','#76B7B2']
# Figure size
plt.figure(figsize=(8,8))

# Donut chart
plt.pie(
    category,
    labels=category.index,
    autopct='%1.1f%%',
    colors=colors,
    shadow=True,
    startangle=90,
    explode=(0.03, 0.03, 0.03)
)

# Create center circle
centre_circle = plt.Circle((0,0), 0.70, fc='white')

# Get current figure
fig = plt.gcf()

# Add circle to chart
fig.gca().add_artist(centre_circle)

# Title
plt.title("Category-wise Sales Distribution", size=18,fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Line Chart For: monthly sales trend

# Convert to datetime format
df["Order Date"]=pd.to_datetime(df['Order Date']) 

# Extract month
#df["Month"]=df['Order Date'].dt.month 

# Convert to Month Name
df['Month Name']= df["Order Date"].dt.month_name()

# Monthly sales
monthly_sales=df.groupby('Month Name')["Sales"].sum()

month_order = ['January', 'February', 'March', 'April',
               'May', 'June', 'July', 'August',
               'September', 'October', 'November', 'December']

monthly_sales = monthly_sales.reindex(month_order)

plt.figure(figsize=(12,5))

# Line chart
plt.plot(monthly_sales.index,
        monthly_sales.values,
        marker='o',
        linewidth=3,
        )

plt.xlabel("Month",size=15,fontweight='bold')
plt.ylabel("Sales",size=15,fontweight='bold')
plt.xticks(rotation=45)
plt.title("Monthly Sales Trend",size=20,fontweight='bold')
plt.grid()
plt.tight_layout()

plt.show()
```

```python
# Histogram For: sales distribution

plt.show()

import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.hist(
    df['Sales'],
    bins=50,
    edgecolor="black"
)

plt.axvline(
    df['Sales'].mean(),
    color="red",
    linestyle="--",
    linewidth=2,
    label="Average Sales"
)

plt.title("Sales Distribution",fontweight='bold')
plt.xlabel("Sales",fontweight='bold')
plt.ylabel("Frequency",fontweight='bold')

plt.grid(True, linestyle='--', alpha=0.5)

plt.legend()

plt.show()
```

```python
plt.style.use('seaborn-v0_8-whitegrid')

# KPI values

total_sales = df['Sales'].sum()

total_profit = df['Profit'].sum()

total_orders = df['Order ID'].nunique()

# Dashboard figure

fig = plt.figure(figsize=(20,16))

# Grid layout

gs = fig.add_gridspec(3,2)

# First subplot

ax1 = fig.add_subplot(gs[0,0])

# Line Chart For: monthly sales trend

# Convert to datetime format

df["Order Date"]=pd.to_datetime(df['Order Date']) 

# Extract month
#df["Month"]=df['Order Date'].dt.month 

# Convert to Month Name

df['Month Name']= df["Order Date"].dt.month_name()

# Monthly sales

monthly_sales=df.groupby('Month Name')["Sales"].sum()

month_order = ['January', 'February', 'March', 'April',
               'May', 'June', 'July', 'August',
               'September', 'October', 'November', 'December']

monthly_sales = monthly_sales.reindex(month_order)

# Line chart

ax1.plot(monthly_sales.index,
        monthly_sales.values,
        marker='o',
        linewidth=3,
        )

# Labels

ax1.set_xlabel("Month",size=15,fontweight='bold')

ax1.set_ylabel("Sales",size=15,fontweight='bold')

ax1.set_title("Monthly Sales Trend",size=20,fontweight='bold')

ax1.tick_params(axis='x', rotation=45)

ax1.grid(True)

# PIE CHART — Category-wise Sales Distribution

ax2 = fig.add_subplot(gs[0,1])

# Category-wise sales

category_sales=df.groupby('Category')["Sales"].sum()

colors=['skyblue', 'orange', 'lightgreen']

ax2.pie(category_sales,
        labels=category_sales.index,
        autopct='%1.2f%%',
        colors=colors,
        shadow=True,
        startangle=90,
        explode=(0.03,0.03,0.09)
       )

#plt.style.use('ggplot')
#plt.style.use('default')
#plt.style.use('dark_background')

# title

ax2.set_title("Category-wise Sales Distribution",size=20,fontweight='bold')

# TOP PRODUCTS CHART

ax3= fig.add_subplot(gs[1,0])

#top 10 products

top_product=df.groupby('Product Name')["Sales"].sum().sort_values(ascending = False).head(10)

colors = ['red', 'blue', 'green', 'orange', 'purple',
          'pink', 'brown', 'cyan', 'magenta', 'yellow']

ax3.barh(top_product.index,top_product.values,color=colors)

ax3.set_title("Top 10 Selling Products",size=18,fontweight='bold')
ax3.set_xlabel("Product Name",size=12,fontweight='bold')
ax3.set_ylabel("Total Sales",size=12,fontweight='bold')
ax3.grid(True,linestyle='--',alpha=0.5)
ax3.tick_params(axis='y', labelsize=6)

# Heatmap:

# Create subplot

ax4 = fig.add_subplot(gs[1,1])

# Correlation matrix
corr = df.corr(numeric_only=True)

# Heatmap

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    ax=ax4
)

# Title
ax4.set_title(
    "Correlation Heatmap",
    size=18,
    fontweight='bold'
)

#loss-making products

ax5= fig.add_subplot(gs[2,0])

#product wise profit
loss_products = df.groupby('Product Name')['Profit'].sum()

# Filter negative profit products
loss_products = loss_products[loss_products < 0]

# Top 10 loss-making products
loss_products = loss_products.sort_values().head(10)

# Horizontal bar chart

plt.barh(
    loss_products.index,
    loss_products.values,
    color='red'
    
 )

ax5.set_xlabel("Profit",size=12,fontweight='bold')
ax5.set_ylabel("Products",size=12,fontweight='bold')

ax5.set_title("Top Loss-Making Products",size=18,fontweight='bold')

ax5.grid(True, linestyle='--', alpha=0.5)

# Smaller product name size
ax5.tick_params(
    axis='y',
    labelsize=8)

# KPI CARDS

# Main Dashboard Title
fig.suptitle("Sales Data Analysis Dashboard",
             fontsize=24,
             fontweight='bold',
             y=1.0)

#KPI 1-> Totla Sales
fig.text(
    0.15,0.93,
    f"Total Sales\n ${total_sales:,.0f}",
    fontsize=16,
    fontweight='bold',
    bbox=dict(
        facecolor='skyblue',
        alpha=0.5,
        boxstyle='round,pad=0.5'
    )
)
# KPI 2 → Total Profit
fig.text(
    0.45, 0.93,
    f"Total Profit\n  ${total_profit:,.0f}",
    fontsize=16,
    fontweight='bold',
    bbox=dict(
        facecolor='lightgreen',
        alpha=0.5,
        boxstyle='round,pad=0.5'
    )
)
# KPI 3 → Total Orders
fig.text(
    0.75, 0.93,
    f"Total Orders\n      {total_orders}",
    fontsize=16,
    fontweight='bold',
    bbox=dict(
        facecolor='orange',
        alpha=0.5,
        boxstyle='round,pad=0.5'
    )
)

plt.subplots_adjust(top= 0.88, hspace= 0.6, wspace= 0.3)

plt.show()
```
